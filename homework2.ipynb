```python
# Problem 3

def sort_and_separate_integers(data):
    data_type = type(data)
    if data_type is not list:
        print("not a list")
        return None
    else:
        print(data)
    even_list = []
    odd_list = []
    for i, value in enumerate(data):
        if not isinstance(value, int):
            print (f"non-integer entry found at index {i}: {data[i]}")
        else:
            if value % 2 == 0:
                even_list.append(value)
            else:
                odd_list.append(value)
    return(sorted(even_list), sorted(odd_list))
```

```python
# Problem 4

def analyze_list_sets(l0, l1):
    if not isinstance(l0, list):
        print (f"invalid input: {l0}")
        return None
    elif not isinstance(l1, list):
        print (f"invalid input: {l1}")
        return None
    primitive_types = (int, float, str, bool, complex, type(None))
    val_l0 = []
    for i, value in enumerate(l0):
        if isinstance(value, primitive_types):
            val_l0.append(value)
        else:
            print (f"invalid entry at index {i} in l0: {l0[i]}")
    val_l1 = []
    for i, value in enumerate(l1):
        if isinstance(value, primitive_types):
            val_l1.append(value)
        else:
            print (f"invalid entry at index {i} in l1: {l1[i]}")

    set_l0 = set(val_l0)
    set_l1 = set(val_l1)


    intersection = sorted(set_l0.intersection(set_l1))
    symmetric_diff = sorted(set_l0.symmetric_difference(set_l1))

    return (intersection, symmetric_diff)
```

```python
analyze_list_sets([1, 'a', 3], [3, 'b', 'a'])
```
