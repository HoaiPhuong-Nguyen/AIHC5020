```python
# Problem 3

def sort_and_separate_integers(data):
    data_type = type(data)
    if data_type is not list:
        print("not a list")
        return None
    else:
        print(data)
    even_list = []
    odd_list = []
    for i, value in enumerate(data):
        if not isinstance(value, int):
            print (f"non-integer entry found at index {i}: {data[i]}")
        else:
            if value % 2 == 0:
                even_list.append(value)
            else:
                odd_list.append(value)
    return(sorted(even_list), sorted(odd_list))
```

```python
# Problem 4

def analyze_list_sets(l0, l1):
    if not isinstance(l0, list):
        print (f"invalid input: {l0}")
        return None
    elif not isinstance(l1, list):
        print (f"invalid input: {l1}")
        return None
    primitive_types = (int, float, str, bool, complex, type(None))
    val_l0 = []
    for i, value in enumerate(l0):
        if isinstance(value, primitive_types):
            val_l0.append(value)
        else:
            print (f"invalid entry at index {i} in l0: {l0[i]}")
    val_l1 = []
    for i, value in enumerate(l1):
        if isinstance(value, primitive_types):
            val_l1.append(value)
        else:
            print (f"invalid entry at index {i} in l1: {l1[i]}")

    set_l0 = set(val_l0)
    set_l1 = set(val_l1)


    intersection = sorted(set_l0.intersection(set_l1), key=lambda x: (str(type(x)), x))
    symmetric_diff = sorted(set_l0.symmetric_difference(set_l1), key=lambda x: (str(type(x)), x))

    return (intersection, symmetric_diff)
```

```python
# Problem 7

import numpy as np

def calculate_regression(x, y):

    if not isinstance(x, np.ndarray) or not isinstance(y, np.ndarray):
        print("invalid input")
        return None

    if x.ndim != 1 or y.ndim != 1:
        print("invalid input")
        return None

    if x.dtype.kind != 'f' or y.dtype.kind != 'f':
        print("invalid input")
        return None

    if x.shape[0] != y.shape[0]:
        print("x and y are not the same length")
        return None

    x_masked = np.ma.masked_invalid(x)
    y_masked = np.ma.masked_invalid(y)

    x_mean = x_masked.mean()
    y_mean = y_masked.mean()

    x_filled = x_masked.filled(x_mean)
    y_filled = y_masked.filled(y_mean)

    numerator = np.sum((x_filled - x_mean) * (y_filled - y_mean))
    denominator = np.sum((x_filled - x_mean) ** 2)

    if denominator == 0:
        print("cannot compute regression: vertical line case")
        return None

    slope = numerator / denominator
    intercept = y_mean - slope * x_mean

    return slope, intercept
```

```python
# Problem 8

import numpy as np
import matplotlib.pyplot as plt

def calculate_regression(x, y):
    if not (isinstance(x, np.ndarray) and isinstance(y, np.ndarray)):
        print("invalid input")
        return None
    if x.ndim != 1 or y.ndim != 1:
        print("invalid input")
        return None
    if not (np.issubdtype(x.dtype, np.floating) and np.issubdtype(y.dtype, np.floating)):
        print("invalid input")
        return None
    if x.shape[0] != y.shape[0]:
        print("x and y are not the same length")
        return None

    x_masked = np.ma.masked_invalid(x)
    y_masked = np.ma.masked_invalid(y)

    x_filled = x_masked.filled(x_masked.mean()).astype(float)
    y_filled = y_masked.filled(y_masked.mean()).astype(float)

    x_mean = x_filled.mean()
    y_mean = y_filled.mean()

    denominator = np.sum((x_filled - x_mean) ** 2)
    if denominator == 0:
        return (np.nan, y_mean)

    numerator = np.sum((x_filled - x_mean) * (y_filled - y_mean))
    slope = numerator / denominator
    intercept = y_mean - slope * x_mean

    return (slope, intercept)

def plot_regression(x, y):
    result = calculate_regression(x, y)
    if result is None:
        return
    slope, intercept = result

    if np.isnan(slope):
        print("vertical line detected")
        return None

    # Impute np.nan values for plotting (safe to reuse masked logic)
    x_imputed = np.ma.masked_invalid(x).filled(np.nanmean(x))
    y_imputed = np.ma.masked_invalid(y).filled(np.nanmean(y))

    # Plotting
    fig, ax = plt.subplots()
    ax.scatter(x_imputed, y_imputed, color='blue', label='Data points')
    ax.set_title("Linear Regression Plot")
    ax.set_xlabel("x")
    ax.set_ylabel("y")
    plt.grid(True)
    plt.grid(color='gray', linestyle='--', linewidth=0.5)
    ax.legend()
    plt.show()
```
