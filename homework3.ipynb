```python
# Problem 6

# Function
def find_top_student_in_major(df, major_name):
  df_filtered = df[df['Major'] == major_name]
  if df_filtered.empty:
        print(f"No students found in the major '{major_name}'.")
        return None
  df_sorted = df_filtered.sort_values(by='Score', ascending=False)
  top_student = df_sorted.iloc[0]
  return top_student[['Name', 'Score']]
```

```python
# Problem 7
stroke_df = pd.read_csv('/content/drive/MyDrive/AIHC5020/stroke/healthcare-dataset-stroke-data.csv')

def determine_cardiovascular_disease(row):
    if row['hypertension'] == 1 or row['heart_disease'] == 1:
        return 1
    else:
        return 0

stroke_df['cardiovascular_disease'] = stroke_df.apply(determine_cardiovascular_disease, axis=1)

print(stroke_df[['hypertension', 'heart_disease', 'cardiovascular_disease']].head())
print("\nValue counts for cardiovascular_disease:")
print(stroke_df['cardiovascular_disease'].value_counts())
```

```python
# Problem 8
import matplotlib.pyplot as plt

def plot_histograms_by_category(df, continuous_col, categorical_col):
    if continuous_col not in df.columns:
        print(f"Error: Column '{continuous_col}' not found in DataFrame.")
        return None
    if categorical_col not in df.columns:
        print(f"Error: Column '{categorical_col}' not found in DataFrame.")
        return None

    n_unique_cont = df[continuous_col].nunique()
    n_unique_cat = df[categorical_col].nunique()

    if n_unique_cont < 10:
        print(f"Error: '{continuous_col}' looks categorical (only {n_unique_cont} unique values).")
        return None
    if n_unique_cat >= 10:
        print(f"Error: '{categorical_col}' looks continuous ({n_unique_cat} unique values).")
        return None

    fig, ax = plt.subplots(figsize=(10, 6))
    unique_categories = df[categorical_col].dropna().unique()

    for category in unique_categories:
        subset = df[df[categorical_col] == category]
        ax.hist(
            subset[continuous_col].dropna(),
            bins=30,
            alpha=0.6,
            label=f"{categorical_col} = {category}"
        )

    ax.set_title(f"Distribution of {continuous_col} by {categorical_col}")
    ax.set_xlabel(continuous_col)
    ax.set_ylabel("Frequency")
    ax.legend()
    plt.show()

plot_histograms_by_category(stroke_df, 'age', 'stroke')

plot_histograms_by_category(stroke_df, 'bmi', 'gender')

plot_histograms_by_category(stroke_df, 'stroke', 'age')
```

```python
# Problem 9

def calculate_stroke_prevalence(df, group_by_cols):
    for col in group_by_cols:
        if col not in df.columns:
            print(f"Error: Column '{col}' not found in DataFrame.")
            return None

        n_unique = df[col].nunique()
        if n_unique >= 10:
            print(f"Error: '{col}' looks continuous ({n_unique} unique values). Expected categorical.")
            return None

    group_sizes = df.groupby(group_by_cols).size()

    stroke_counts = df.groupby(group_by_cols)['stroke'].sum()

    prevalence = stroke_counts / group_sizes

    result = prevalence.reset_index(name="stroke_prevalence")

    return result

print("\nPrevalence by smoking_status:")
print(calculate_stroke_prevalence(stroke_df, ['smoking_status']))

print("\nPrevalence by ever_married and work_type:")
print(calculate_stroke_prevalence(stroke_df, ['ever_married', 'work_type']))

print("\nPrevalence by age:")
print(calculate_stroke_prevalence(stroke_df, ['age']))

print("\nPrevalence by residence_type:")
print(calculate_stroke_prevalence(stroke_df, ['residence_type']))
```

```python
# Problem 10

def preprocess_for_analysis(df, cols_to_impute, cols_to_scale, cols_to_encode, cols_to_drop):
    df_processed = df.copy()

    for col in cols_to_drop:
        if col in df_processed.columns:
            df_processed.drop(columns=col, inplace=True)
        else:
            print(f"Warning: Column '{col}' not found. Skipping drop.")

    for col in cols_to_impute:
        if col in df_processed.columns:
            if pd.api.types.is_numeric_dtype(df_processed[col]):
                mean_value = df_processed[col].mean()
                df_processed[col].fillna(mean_value, inplace=True)
            else:
                print(f"Warning: Column '{col}' is not numeric. Skipping imputation.")
        else:
            print(f"Warning: Column '{col}' not found for imputation.")

    for col in cols_to_scale:
        if col in df_processed.columns:
            if pd.api.types.is_numeric_dtype(df_processed[col]):
                mean_val = df_processed[col].mean()
                std_val = df_processed[col].std()
                if std_val == 0:
                    print(f"Warning: Column '{col}' has zero std deviation. Skipping scaling.")
                    continue
                df_processed[f"{col}_scaled"] = (df_processed[col] - mean_val) / std_val
                df_processed.drop(columns=col, inplace=True)
            else:
                print(f"Warning: Column '{col}' is not numeric. Skipping scaling.")
        else:
            print(f"Warning: Column '{col}' not found for scaling.")

    for col in cols_to_encode:
        if col in df_processed.columns:
            if not pd.api.types.is_numeric_dtype(df_processed[col]):
                encoded = pd.get_dummies(df_processed[col], prefix=col, drop_first=True)
                df_processed.drop(columns=col, inplace=True)
                df_processed = pd.concat([df_processed, encoded], axis=1)
            else:
                print(f"Warning: Column '{col}' is numeric. Skipping encoding.")
        else:
            print(f"Warning: Column '{col}' not found for encoding.")

    return df_processed

scale_list = ['age', 'avg_glucose_level', 'bmi']
encode_list = ['gender', 'ever_married', 'work_type', 'Residence_type', 'smoking_status']
drop_list = ['id', 'location_code']
impute_list = ['bmi']

processed_df = preprocess_for_analysis(
    stroke_df,
    cols_to_impute=impute_list,
    cols_to_scale=scale_list,
    cols_to_encode=encode_list,
    cols_to_drop=drop_list
)

print(processed_df.head())
print(processed_df.info())
```
